Simulated Annealing

```python
import math
import random


def calculate_conflicts(state):
    """Calculates the number of attacking pairs of queens (Objective Function).

    A score of 0 means a perfect solution has been found.
    """
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Check row conflict or diagonal conflict
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def get_neighbor(state):
    """Generates a neighbor state by randomly picking one column

    and changing its queen's row position.
    """
    neighbor = list(state)
    col = random.randint(0, 7)
    current_row = state[col]

    # Pick a different row for this column
    available_rows = [r for r in range(8) if r != current_row]
    neighbor[col] = random.choice(available_rows)

    return neighbor


def simulated_annealing(
    initial_temp=100.0, cooling_rate=0.99, min_temp=0.001, max_iter=10000
):
    """Solves the 8-Queens problem using Simulated Annealing."""
    # Generate a random initial state (1 queen per column)
    current_state = [random.randint(0, 7) for _ in range(8)]
    initial_state = list(current_state)
    current_energy = calculate_conflicts(current_state)

    temp = initial_temp
    iteration = 0

    print(f"Initial board conflicts: {current_energy}")
    print(f"Initial array representation: {initial_state}")
    print("Initial board configuration:")
    print_board(initial_state)
    print("\nOptimizing...\n")

    while temp > min_temp and current_energy > 0 and iteration < max_iter:
        neighbor = get_neighbor(current_state)
        neighbor_energy = calculate_conflicts(neighbor)

        # Energy difference (we want to minimize conflicts, so lowering energy is good)
        delta_e = neighbor_energy - current_energy

        # If neighbor is better, or probabilistically accept a worse move to escape local minima
        if delta_e < 0 or random.random() < math.exp(-delta_e / temp):
            current_state = neighbor
            current_energy = neighbor_energy

        # Cool down the temperature
        temp *= cooling_rate
        iteration += 1

    return initial_state, current_state, current_energy, iteration


def print_board(state):
    """Visualizes the board configuration cleanly."""
    for row in range(8):
        line = ""
        for col in range(8):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)


# --- Execution ---
if __name__ == "__main__":
    # Seed for reproducibility (optional)
    # random.seed(42)

    init_state, final_state, final_conflicts, total_iters = simulated_annealing(
        initial_temp=100.0, cooling_rate=0.99, min_temp=0.001, max_iter=10000
    )

    print("--- Execution Complete ---")
    print(f"Total iterations: {total_iters}")
    print(f"Final conflicts: {final_conflicts}")
    print(f"Final array representation: {final_state}\n")

    print("Final board configuration:")
    print_board(final_state)
    print()

    if final_conflicts == 0:
        print("Success! Found a valid solution.")
    else:
        print("Could not find a perfect solution. Try adjusting parameters.")
```
