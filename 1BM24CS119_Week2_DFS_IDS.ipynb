DFS

```python
# Import necessary libraries
from collections import deque

# Define the dimensions of the puzzle
N = 3

# Structure to store a state of the puzzle
class PuzzleState:
    def __init__(self, board, x, y, depth):
        self.board = board
        self.x = x
        self.y = y
        self.depth = depth

# Possible moves: Left, Right, Up, Down
row = [0, 0, -1, 1]
col = [-1, 1, 0, 0]

# Function to check if a given state is the goal state
def is_goal_state(current_board, goal_board):
    return current_board == goal_board

# Function to check if a move is valid
def is_valid(x, y):
    return 0 <= x < N and 0 <= y < N

# Function to print the board
def print_board(board):
    for row in board:
        print(' '.join(map(str, row)))
    print("--------")

# Depth-First Search to solve the 8-puzzle problem
def solve_puzzle_dfs(start, x, y, end_board):
    stack = []
    visited = set()

    stack.append(PuzzleState(start, x, y, 0))
    visited.add(tuple(map(tuple, start)))

    while stack:
        curr = stack.pop()

        # Print the current board
        print(f'Depth: {curr.depth}')
        print_board(curr.board)

        # Check if goal state is reached
        if is_goal_state(curr.board, end_board):
            print(f'Goal state reached at depth {curr.depth}')
            return

        # Explore possible moves
        for i in range(4):
            new_x = curr.x + row[i]
            new_y = curr.y + col[i]

            if is_valid(new_x, new_y):
                new_board = [row[:] for row in curr.board]
                # Swap the tiles
                new_board[curr.x][curr.y], new_board[new_x][new_y] = new_board[new_x][new_y], new_board[curr.x][curr.y]

                # If this state has not been visited before, push to stack
                board_tuple = tuple(map(tuple, new_board))
                if board_tuple not in visited:
                    visited.add(board_tuple)
                    stack.append(PuzzleState(new_board, new_x, new_y, curr.depth + 1))

    print('No solution found (DFS Brute Force reached depth limit)')

# Driver Code
if __name__ == '__main__':
    start_board = [[2,8,3], [1,6,4], [7,0,5]]
    start_x, start_y = 1, 1 # Position of the empty tile (0)
    end_board = [[1, 2, 3],[8,0,4],[7,6,5]]

    print('Initial State:')
    print_board(start_board)

    solve_puzzle_dfs(start_board, start_x, start_y, end_board)
```

IDDFS

```python
# Goal state configuration
GOAL_STATE = (1, 2, 3,
              8, 0, 4,
              7, 6, 5)

# Moves map based on the 0-indexed position of the blank tile (0)
# Format: index -> list of indices it can swap with (Up, Down, Left, Right)
MOVES = {
    0: [1, 3],       1: [0, 2, 4],    2: [1, 5],
    3: [0, 4, 6],    4: [1, 3, 5, 7], 5: [2, 4, 8],
    6: [3, 7],       7: [4, 6, 8],    8: [5, 7]
}

def get_neighbors(state):
    """Generates next valid states by moving the blank tile (0)."""
    neighbors = []
    blank_idx = state.index(0)

    for next_idx in MOVES[blank_idx]:
        # Swap blank tile with adjacent tile
        new_state = list(state)
        new_state[blank_idx], new_state[next_idx] = new_state[next_idx], new_state[blank_idx]
        neighbors.append(tuple(new_state))

    return neighbors

def depth_limited_search(state, depth, path, visited_path):
    """Performs a depth-limited DFS tracking the current recursion path."""
    if state == GOAL_STATE:
        return path

    if depth <= 0:
        return None

    # Add to current path to prevent cycles in this branch
    visited_path.add(state)

    for neighbor in get_neighbors(state):
        if neighbor not in visited_path:
            result = depth_limited_search(neighbor, depth - 1, path + [neighbor], visited_path)
            if result is not None:
                return result

    # Backtrack
    visited_path.remove(state)
    return None

def iddfs(initial_state, max_depth=3):
    """Executes Iterative Deepening DFS up to a maximum depth limit."""
    print("Searching for solution...")
    for depth in range(max_depth + 1):
        # Tracking states along the current recursion branch to avoid loops
        visited_path = set()
        result = depth_limited_search(initial_state, depth, [initial_state], visited_path)

        if result is not None:
            return result, depth

    return None, max_depth

def print_board(state):
    """Helper to format and print the 3x3 puzzle board."""
    for i in range(0, 9, 3):
        print(f" {state[i]} {state[i+1]} {state[i+2]} ")
    print("-" * 11)

# --- Execution ---
if __name__ == "__main__":
    # A solvable puzzle configuration (Requires 4 moves)
    initial_puzzle = (2, 8, 3,
                      1, 6, 4,
                      7, 0, 5)

    solution_path, depth_found = iddfs(initial_puzzle)

    if solution_path:
        print(f"\nSuccess! Solution found at Depth: {depth_found}")
        print("Steps:")
        for step, board in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(board)
    else:
        print("No solution found within depth limits.")
```
